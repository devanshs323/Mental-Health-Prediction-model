Ordinal Encoding from sklearn


```python
import pandas as pd
df1=pd.DataFrame({"size":["s","m","l","xl","m","l","xl","s","m","l","xl","s","m","xl","s","l","xl" ]})
ord_data=[["s",'m','l','xl']]
from sklearn.preprocessing import OrdinalEncoder
ord_enc=OrdinalEncoder(categories=ord_data)
df1['Encoded'] = ord_enc.fit_transform(df1[['size']])
df1
```

Ordinal Encoding by Map Function


```python
ord_data1={"s":1,"m":2,"l":3,"xl":4}
df1['Encoded1']=df1['size'].map(ord_data1)
df1
```
