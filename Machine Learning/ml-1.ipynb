```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel(r'C:\Users\devan\Downloads\ml_data_cleaning_practice.xlsx')
df.head()
df.isnull().sum()
```

```python
(df.isnull().sum()/df.shape[0]) * 100
plt.figure(figsize=(10,5))
plt.plot( df.columns, (df.isnull().sum()/df.shape[0]) * 100, marker='o', linestyle='-', color='b')
plt.xlabel('Columns')
plt.ylabel('Percentage of null values')
plt.title('Null Value Analysis')
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
null_val= (df.isnull().sum().sum()/(df.shape[0] * df.shape[1]) * 100)
not_null_val = 100 - null_val
null1= df.isnull().sum().sum()
not_null1 = df.notnull().sum().sum()        
print('Percentage of null values in the dataset is: ',null1,null_val)
print('Percentage of not null values in the dataset is: ', not_null1,not_null_val)

```
