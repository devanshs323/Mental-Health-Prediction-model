checking outliers

```python
num_features = df.select_dtypes(include='number')
Q1=num_features.quantile(0.25)
Q3=num_features.quantile(0.75)
IQR=Q3-Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
outliers = (num_features<lower_bound) | (num_features>upper_bound)
outliers.sum()


```

```python
df.drop_duplicates()

# removing negative or unrealistic data to realistic or 0
df['Physical_Activity_Hours']=df['Physical_Activity_Hours'].clip(lower=0)
df.describe()
```

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
top_countries
```

```python
def  group_countries(country):
    if(country) in top_countries:
        return country
    else:
        return 'Other'
df['Grouped_country']= df['Country'].apply(group_countries)
    
```

```python
df['Grouped_country'].value_counts()
```

```python
num_cols = df.select_dtypes(include='number')
num_cols.skew() #study_hours
df.columns
```

```python
from sklearn.model_selection import train_test_split


skewd_col = ['Study_Hours']
other_num_col = ['Avg_Daily_Usage_Hours', 'Daily_Unlocks','Age','Physical_Activity_Hours', 'Sleep_Hours_Per_Night']
ordinal_col = ['Stress_Level']
nominal_col = ['Gender', 'Country', 'Academic_Level', 'Most_Used_Platform','Purpose_Of_Use','Grouped_country']

feature_col = skewd_col + other_num_col + ordinal_col + nominal_col 
x=df[feature_col]
y=df['Mental_Health_Score']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer
# skewed Features
skewed_pipeline= Pipeline(steps=[
    ('log_transform',FunctionTransformer(np.log1p)),
    ('scale',StandardScaler())
])

#numeric features
plain_numeric_pipeline= Pipeline(steps=[
    ('scale',StandardScaler())

])
# ordinal features 
ordinal_pipeline= Pipeline(steps=[
    ('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])

#  nominal features
nominal_pipeline= Pipeline(steps=[
    ('encode',OneHotEncoder(handle_unknown='ignore'))
])

preprocessor =  ColumnTransformer(transformers=[
    ("skewed_pipeline", skewed_pipeline,skewd_col),
    ("plain_numeric_pipeline",plain_numeric_pipeline,other_num_col),
    ("ordinal_pipeline",ordinal_pipeline,ordinal_col),
    ("nominal_pipeline",nominal_pipeline,nominal_col)
])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error,mean_absolute_error
lr_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor',LinearRegression())
]) 

lr_pipeline.fit(x_train,y_train)
```

```python
# if the diffrence b/w both the accuracy it means there is the chances of overfitting and underfitting

lr_pred = lr_pipeline.predict(x_test)
lr_pred_train = lr_pipeline.predict(x_train)

lr_r2_train = r2_score(y_train,lr_pred_train)
lr_r2 = r2_score(y_test,lr_pred)
mae= mean_absolute_error(y_test,lr_pred)

print(lr_r2_train)
print(lr_r2)
print(mae)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline = Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('random_forest',RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(x_train,y_train)
```

```python
rf_pred = rf_pipeline.predict(x_test)
rf_pred_train = rf_pipeline.predict(x_train)

rf_r2_train = r2_score(y_train,rf_pred_train)
rf_r2 = r2_score(y_test,rf_pred)
rf_mae= mean_absolute_error(y_test,rf_pred)

print(rf_r2_train)
print(rf_r2)
print(rf_mae)
```

Hyperparameter tunning


```python
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'random_forest__n_estimators': [100, 200, 300],
    'random_forest__max_depth': [5, 10, 15],
    'random_forest__min_samples_split': [2, 5, 10],
    'random_forest__min_samples_leaf': [1, 2, 4],
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_grid,
    cv=5,
    scoring='r2',
    random_state=42,
    n_iter=15,
    n_jobs=-1,
)

random_search.fit(x_train, y_train)
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_pred = rf_best_pipeline.predict(x_test)
print(r2_score(y_test,rf_best_pred))
print(mean_absolute_error(y_test,rf_best_pred))
```

```python
from sklearn.metrics import mean_squared_error

# Calculate RMSE for Linear Regression
lr_rmse = np.sqrt(mean_squared_error(y_test, lr_pred))

# Calculate RMSE for default Random Forest
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))

# Calculate RMSE for tuned Random Forest
rf_tuned_preds           = random_search.best_estimator_.predict(x_test)
rf_tuned_rmse            = np.sqrt(mean_squared_error(y_test, rf_tuned_preds))
rf_tuned_training_preds  = random_search.best_estimator_.predict(x_train)
r2_tuned_training        = r2_score(y_train, rf_tuned_training_preds)
rf_tuned_mae             = mean_absolute_error(y_test, rf_tuned_preds)
rf_tuned_r2              = r2_score(y_test, rf_tuned_preds)

# Create a DataFrame to consolidate results
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest (default)', 'Random Forest (tuned)'],
    'R2': [lr_r2, rf_r2, rf_tuned_r2],
    'Training R2': [lr_r2_train, rf_r2_train, r2_tuned_training],
    'MAE': [mae, rf_mae, rf_tuned_mae],
    'RMSE': [lr_rmse, rf_rmse, rf_tuned_rmse]
})

print(results)
```
